```python
# lets make grading system
# if scores are >= 90 print A grade
# if scores are >= 80 print B grade
# if scores are >= 70 print C grade
# if scores are >= 60 print D grade
```

```python
score=int(input("Enter your score"))
if score >= 90:
    print("Grade A. Congratulations")

elif score >=80:
    print("Grade B. Better luck next time")

elif score >=70:
    print("Grade C. Study more")

elif score >=60:
    print("Grade D. Thodkyat vachla")

else:
    print("Fail. Shala sodun ghari ja")
```
